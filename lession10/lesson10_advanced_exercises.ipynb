## Exercise 1: Array statistics

- Create numeric array/list with at least 12 elements.
- Compute min, max, sum, average, and median (manual or built-in helper).
- Print all metrics.

```python
# Write your solution here
numbers = [42, 17, 29, 88, 53, 61, 25, 73, 36, 92, 48, 65]

minimum = min(numbers)
maximum = max(numbers)
total = sum(numbers)
average = total / len(numbers)
ordered = sorted(numbers)
middle = len(ordered) // 2
if len(ordered) % 2 == 0:
    median = (ordered[middle - 1] + ordered[middle]) / 2
else:
    median = ordered[middle]

print("Min:", minimum)
print("Max:", maximum)
print("Sum:", total)
print("Average:", average)
print("Median:", median)
```

## Exercise 2: Iterator protocol deep dive

- Create iterator from a string and consume values with `next()` until exhausted.
- Handle `StopIteration` explicitly.
- Print total number of consumed characters.

```python
# Write your solution here
text = "Python"
text_iterator = iter(text)
consumed = 0

while True:
    try:
        character = next(text_iterator)
        print(character)
        consumed += 1
    except StopIteration:
        print("Iterator exhausted")
        break

print("Total consumed characters:", consumed)
```

## Exercise 3: Custom iterator class

- Implement class `EvenNumbers(limit)` yielding even numbers up to limit.
- Use `__iter__` and `__next__`.
- Iterate and print values.

```python
# Write your solution here
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.limit:
            raise StopIteration
        value = self.current
        self.current += 2
        return value

for number in EvenNumbers(20):
    print(number)
```

## Exercise 4: Iterator state with `next()`

- Create a custom iterator class `Countdown(start)` that counts down to 1.
- Use `iter()` and `next()` to take the first two values manually.
- Then use a `for` loop on the same iterator to get the remaining values.
- Add a short comment explaining why the `for` loop continues where `next()` stopped.

```python
# Write your solution here
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < 1:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

countdown = Countdown(5)
print("First:", next(iter(countdown)))
print("Second:", next(countdown))

# The for loop continues using the same iterator object and its remaining internal state.
for number in countdown:
    print(number)
```
