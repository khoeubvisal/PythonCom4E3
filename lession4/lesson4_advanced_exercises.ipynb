## Exercise 1: Nested list processing

- Create a 2D list (3 rows x 4 cols) of numbers.
- Print row sums and grand total.
- Print the maximum value from each row.

```python
# Write your solution here
# 1. Create a 2D list (3 rows x 4 cols) of numbers
matrix = [
    [15, 25, 35, 8],
    [32, 19, 45, 12],
    [7, 29, 18, 36]
]

grand_total = 0

# 2 & 3. Print row sums, grand total, and maximum value from each row
for i, row in enumerate(matrix, start=1):
    row_sum = sum(row)
    row_max = max(row)
    grand_total += row_sum
    print(f"Row {i}: Sum = {row_sum}, Max = {row_max}")

print("Grand Total:", grand_total)
```

## Exercise 2: Comprehension with condition

- From numbers 1..30, build a list of numbers divisible by 3 but not by 2.
- Then build a second list of their squares using comprehension.
- Print both lists.

```python
# Write your solution here
# 1. Numbers from 1 to 30 divisible by 3 but not by 2
filtered_numbers = [n for n in range(1, 31) if n % 3 == 0 and n % 2 != 0]

# 2. Second list of their squares using comprehension
squares = [n ** 2 for n in filtered_numbers]

# 3. Print both lists
print("Filtered numbers:", filtered_numbers)
print("Squares:", squares)
```

## Exercise 3: Custom sorting

- Create a list of lists: `[name, score]` for at least 5 students.
- Write a helper function that returns a sort key `[-score, name]`.
- Sort descending by score, then ascending by name for ties, using `sort(key=...)`.
- Print the sorted result.

```python
# Write your solution here
# 1. Create a list of lists: [name, score] for at least 5 students
students = [
    ["Sal", 85],
    ["sok", 92],
    ["Bob", 85],
    ["lin", 78],
    ["Nita", 92]
]

# 2. Helper function that returns a sort key [-score, name]
def sort_key(student):
    name, score = student
    return [-score, name]

# 3. Sort descending by score, then ascending by name for ties
students.sort(key=sort_key)

# 4. Print the sorted result
for name, score in students:
    print(f"{name}: {score}")
```

## Exercise 4: Duplicate removal (order preserved)

- Given a list with duplicates, create a new list without duplicates while preserving original order.
- Use a loop and a membership check (`not in`) before appending.
- Print original and cleaned lists.

```python
# Write your solution here
# Given a list with duplicates
original_list = [4, 2, 7, 2, 4, 9, 1, 7, 3]
cleaned_list = []

# Use a loop and a membership check (not in) before appending
for item in original_list:
    if item not in cleaned_list:
        cleaned_list.append(item)

# Print original and cleaned lists
print("Original list:", original_list)
print("Cleaned list:", cleaned_list)
```

## Exercise 5: Merge and flatten

- Create two lists of lists, merge them, and flatten into one list.
- Print merged nested list and flattened list.
- Compute and print average of flattened numeric values.

```python
# Write your solution here
# 1. Create two lists of lists and merge them
list1 = [[10, 20], [30, 40]]
list2 = [[50, 60], [70, 80]]

merged_nested = list1 + list2

# Flatten into one list
flattened = [num for sublist in merged_nested for num in sublist]

# 2. Print merged nested list and flattened list
print("Merged nested list:", merged_nested)
print("Flattened list:", flattened)

# 3. Compute and print average of flattened numeric values
average = sum(flattened) / len(flattened)
print("Average:", average)
```
