## Exercise 1: Tuple unpacking patterns

- Create a tuple with at least 7 values.
- Unpack into `first`, `second`, `*middle`, `last`.

```python
# Write your solution here
values = (10, 20, 30, 40, 50, 60, 70)
first, second, *middle, last = values

print("First:", first)
print("Second:", second)
print("Middle:", middle)
print("Middle length:", len(middle))
print("Last:", last)
```

## Exercise 2: Tuple transformation

- Start from a tuple of integers.
- Create a new tuple with each value multiplied by 10.
- Print original and transformed tuples.

```python
# Write your solution here
# 1. Start from a tuple of integers
original_tuple = (2, 4, 6, 8, 10)

# 2. Create a new tuple with each value multiplied by 10
transformed_tuple = tuple(num * 10 for num in original_tuple)

# 3. Print original and transformed tuples
print("Original tuple:", original_tuple)
print("Transformed tuple:", transformed_tuple)
```

## Exercise 3: Set-based analytics

- Given two sets of course names, print union, intersection, and both differences.
- Print symmetric difference and explain what it means in a comment.

```python
# Write your solution here
python_course = {"Python", "Data Science", "Machine Learning", "SQL"}
web_course = {"HTML", "CSS", "Python", "JavaScript", "SQL"}

print("Union:", python_course | web_course)
print("Intersection:", python_course & web_course)
print("Python course only:", python_course - web_course)
print("Web course only:", web_course - python_course)

symmetric_difference = python_course ^ web_course
print("Symmetric difference:", symmetric_difference)
# It contains courses that are in exactly one of the two sets.
```

## Exercise 4: Set cleanup task

- Create a set from a list containing repeated and mixed-case words.
- Normalize words to lowercase before inserting into set.
- Print final normalized unique set sorted alphabetically.

```python
# Write your solution here
words = ["Apple", "banana", "APPLE", "Cherry", "Banana", "cherry", "Pear"]
normalized_words = {word.lower() for word in words}

print("Normalized unique set sorted alphabetically:", sorted(normalized_words))
```

## Exercise 5: Set from a list comprehension

- Use a list comprehension to build the squares of numbers 1..20 where the square is even.
- Convert the list to a set to remove duplicates (if any).
- Print the resulting set and its length.
- Convert it to a sorted list and print.

```python
# Write your solution here
even_squares_list = [number ** 2 for number in range(1, 21) if (number ** 2) % 2 == 0]
even_squares_set = set(even_squares_list)

print("Resulting set:", even_squares_set)
print("Set length:", len(even_squares_set))
print("Sorted list:", sorted(even_squares_set))
```
